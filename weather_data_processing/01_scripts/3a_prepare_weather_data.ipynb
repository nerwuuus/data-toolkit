### Prepare Raw Weather Data

This notebook combines all raw NOAA weather observation files into a single dataset, performs basic data preparation, and exports the processed data in **CSV** and **Apache Parquet** formats for downstream processing and analysis.

```python
# Import libraries
from pathlib import Path
import pandas as pd
import pyarrow as pa
```

### Read Raw Files

```python
# Read the CSV/Excel files into DataFrames
raw_path = Path(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw_data"
)

dfs = []

for file in raw_path.glob("*.csv.gz"):
    df = pd.read_csv(
        file, 
        header=None
    )
    dfs.append(df)

weather = pd.concat(
    dfs,
    ignore_index=True
)
```

### Rename Columns

```python
weather.head()
```

```python
weather.rename(
    columns={
        0:'station',
        1:'observation_date',
        2:'metric',
        3:'value',
        4:'measurement_flag',
        5:'quality_flag',
        6:'source_flag',
        7:'observation_time'
        },
        inplace=True
    )
```

```python
weather
```

### Save as CSV file

```python
weather.to_csv(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw_data\weather.csv",
    index=False
)
```

### Save as a Parquet file

```python
weather.to_parquet(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw_data\weather.parquet",
    index=False,
    engine='pyarrow', # Parquet supports multiple engines; pyarrow is the most commonly used
    compression='snappy' # Compresses the file while keeping it fast to read and write. Compression saves storage and speeds up processing
)
```
