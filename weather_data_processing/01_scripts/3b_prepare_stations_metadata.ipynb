```python
# ============================================================================
# Extract and clean NOAA GHCN weather station metadata.
# Reads the fixed-width source file, converts it into a structured dataset,
# and exports the result as a CSV file for downstream processing.
# ============================================================================
```

```python
import pandas as pd
```

```python
url = "https://www.ncei.noaa.gov/pub/data/ghcn/daily/ghcnd-stations.txt"

stations = pd.read_fwf( # Fixed-width formatted file
    url,
    # Read the fixed-width station metadata file using column positions
    # defined in the NOAA GHCN-D file specification.
    colspecs=[
        (0, 11),
        (12, 20),
        (21, 30),
        (31, 37),
        (38, 40),
        (41, 71),
        (72, 75),
        (76, 79),
        (80, 85)
    ],
    names=[
        "station",
        "latitude",
        "longitude",
        "elevation",
        "state",
        "station_name",
        "gsn_flag",
        "hcn_flag",
        "wmo_id"
    ]
)
```

```python
stations.to_csv(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00 raw\ghcnd-stations_clean.csv",
    index=False
)
```
