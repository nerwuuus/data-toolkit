### Prepare Raw Weather Data Using Polars

This notebook combines all raw NOAA weather observation files into a single dataset, performs basic data preparation, and exports the processed data in **CSV** and **Apache Parquet** formats for downstream processing and analysis.

```python
# Import libraries and define paths
from pathlib import Path
import polars as pl
import glob

RAW_DATA_PATH = Path(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw_data"
)

CSV_EXPORT = RAW_DATA_PATH / "weather_polars.csv"
PARQUET_EXPORT = RAW_DATA_PATH / "weather_polars.parquet"
```

### Read Raw Files

```python
# Read and combine all compressed CSV files
```

```python
weather = pl.read_csv(
    RAW_DATA_PATH / "*.csv.gz",
    has_header=False
)

weather
```

### Rename Columns

```python
weather.head()
```

```python
weather = weather.rename({
    "column_1": "station",
    "column_2": "observation_date",
    "column_3": "metric",
    "column_4": "value",
    "column_5": "measurement_flag",
    "column_6": "quality_flag",
    "column_7": "source_flag",
    "column_8": "observation_time"
})
```

```python
weather.head()
```

### Save as CSV file

```python
weather.write_csv(CSV_EXPORT)
```

### Save as a Parquet file

```python
weather.write_parquet(
    PARQUET_EXPORT,
    compression='snappy'
    )
```
