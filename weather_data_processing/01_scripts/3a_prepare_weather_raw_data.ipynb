```python
# ============================================================================
# Extract and consolidate raw weather observations.
# Loads all compressed CSV files from the raw layer, combines them into a
# single DataFrame, and stores the result as a flat CSV file for downstream
# transformation and analysis.
# ============================================================================
```

```python
# Import libraries
from pathlib import Path
import pandas as pd
```

```python
# Read the CSV/Excel files into DataFrames
raw_path = Path(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw"
)

dfs = []

for file in raw_path.glob("*.csv.gz"):
    df = pd.read_csv(
        file, 
        header=None
    )
    dfs.append(df)

weather = pd.concat(
    dfs,
    ignore_index=True
)
```

```python
weather.to_csv(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw\weather.csv",
    index=False
)
```
