### Prepare Sample weather_sample Data

This notebook creates a 100 000-row sample from the raw NOAA weather_sample observation data, performs basic data preparation, and exports the sample as a CSV file for testing the Docker PostgreSQL loading process.

```python
# Import libraries and define paths
from pathlib import Path
import pandas as pd

RAW_DATA_PATH = Path(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw_data"
)

CSV_EXPORT = RAW_DATA_PATH / "weather_sample.csv"
```

### Read 100 000 Sample Rows

```python
# Read 100 000 sample rows
weather_sample = pd.read_csv(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00_raw_data\2025.csv.gz", 
    header=None,
    nrows=100_000
)
```

### Rename Columns

```python
weather_sample.head()
```

```python
weather_sample.rename(
    columns={
        0:'station',
        1:'observation_date',
        2:'metric',
        3:'value',
        4:'measurement_flag',
        5:'quality_flag',
        6:'source_flag',
        7:'observation_time'
        },
        inplace=True
    )
```

```python
weather_sample.head()
```

### Save as CSV file

```python
weather_sample.to_csv(
    CSV_EXPORT,
    index=False
)
```
