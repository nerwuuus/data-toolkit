```python
# Import libraries
from pathlib import Path
import pandas as pd
import numpy as np 
```

```python
# Read the CSV/Excel files into DataFrames
raw_path = Path(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00 raw"
)

dfs = []

for file in raw_path.glob("*.csv.gz"):
    df = pd.read_csv(
        file, 
        header=None
    )
    dfs.append(df)

weather = pd.concat(
    dfs,
    ignore_index=True
)
```

```python
weather.to_csv(
    r"C:\Users\zychl\Desktop\Data Engineering\weather_data_processing\00 raw\weather.csv",
    index=False
)
```
